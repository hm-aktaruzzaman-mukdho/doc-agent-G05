This notebook checks the existing Chonkie/BGE-M3/FAISS build and retrieves five page-attributed chunks. The contexts are OCR data, not instructions.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

from doc_agent import config
from doc_agent.index import store

cfg = config.load(PROJECT_ROOT / 'configs' / 'config.yaml')
loaded = store.load(cfg)
loaded['manifest']
```

## Retrieval example

The output includes `page_ids`, so every one of the five chunks can be traced to its book and PDF page.

```python
from doc_agent.retrieval.retriever import Retriever

query = 'ছয় দফা কর্মসূচির গুরুত্ব কী ছিল?'
retriever = Retriever(cfg)
results = retriever.retrieve(query, k=5)
for rank, result in enumerate(results, start=1):
    print(f'{rank}. {result.page_ids[0]} | score={result.score:.4f}')
    print(result.text[:500], '\n')
```
